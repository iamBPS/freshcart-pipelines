```python
import os
import subprocess
from pathlib import Path

# Fix Mage environment configuration
os.environ["DEBUG"] = "0"

# Locate the actual project root
current = Path.cwd()

while current != current.parent:
    if (current / "iris_project" / "pipelines").is_dir():
        project_root = current
        break
    current = current.parent
else:
    raise FileNotFoundError("Could not locate iris_project.")

print("Project root:", project_root)

# Run Mage from the correct directory
command = [
    r"C:\Users\hp\AppData\Roaming\Python\Python312\Scripts\mage.exe",
    "run",
    "iris_project",
    "iris_classifier",
    "--test"
]

result = subprocess.run(
    command,
    capture_output=True,
    text=True,
    cwd=str(project_root)
)

print(result.stdout)
print(result.stderr)

if result.returncode != 0:
    raise RuntimeError("Mage pipeline failed. Check the output above.")

print("Mage pipeline completed successfully.")
```
